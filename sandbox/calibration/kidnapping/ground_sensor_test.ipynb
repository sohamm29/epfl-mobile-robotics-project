# Ground sensor test (kidnapping threshold)

Prints the two ground sensor values (`prox.ground.delta`, the same values the state machine uses) for `DURATION` seconds.
The motors stay off: move the robot by hand while it runs.

Try: on the orange map, on the black border / a tag, and lifted. Then choose `ground_threshold` in `src/config.yaml`
below the black values and above the lifted values.

```python
DURATION = 10.0  # seconds
PERIOD = 0.1  # seconds between readings (same as the control loop)
```

```python
from tdmclient import ClientAsync

with ClientAsync() as client:
    node = await client.wait_for_node(timeout=10)
    if node is None:
        raise TimeoutError("No Thymio found. Check TDM and the robot connection.")

    with await node.lock():
        await node.wait_for_variables({"prox.ground.delta"})
        for i in range(round(DURATION / PERIOD)):
            left, right = node.v.prox.ground.delta
            print(f"t={i * PERIOD:4.1f}s  left={left:4d}  right={right:4d}")
            await client.sleep(PERIOD)
```
