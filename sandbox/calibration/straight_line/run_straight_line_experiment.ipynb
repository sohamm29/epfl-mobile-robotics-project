# Straight-line test

With the Thymio on a clear, flat surface and TDM running, run the code cell once. It sets both motor targets to 50 for 2 seconds, then sets both to 0. Measure and write down the distance yourself. Keep the robot in sight and be ready to stop it if the connection fails.

```python
from tdmclient import ClientAsync

with ClientAsync() as client:
    node = await client.wait_for_node(timeout=10)
    if node is None:
        raise TimeoutError("No Thymio found. Check TDM and the robot connection.")

    with await node.lock():
        try:
            await node.set_variables(
                {
                    "motor.left.target": [200],
                    "motor.right.target": [200],
                }
            )
            await client.sleep(5)
        finally:
            await node.set_variables(
                {
                    "motor.left.target": [0],
                    "motor.right.target": [0],
                }
            )

print("Motors stopped. Measure and record the distance.")
```
